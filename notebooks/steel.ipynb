## KPI Automation

```python
total_production = kpi['Production_Tonnes'].sum()

avg_yield = kpi['Yield_Percent'].mean()

avg_achievement = kpi['Achievement_Percent'].mean()

total_revenue = kpi['Revenue_INR'].sum()

total_profit = (kpi['Production_Tonnes'] * kpi['Profit_per_Tonne_INR']).sum()

avg_energy = kpi['Energy_Rate_kWh_per_Tonne'].mean()

summary = pd.DataFrame({
    'Metric': [
        'Total Production',
        'Average Yield',
        'Average Achievement',
        'Total Revenue',
        'Total Profit',
        'Average Energy Rate'
    ],
    'Value': [
        round(total_production,2),
        round(avg_yield,2),
        round(avg_achievement,2),
        round(total_revenue,2),
        round(total_profit,2),
        round(avg_energy,2)
    ]
})

summary
```

## Plant Condition 

```python
kpi['Plant_Health_Score'] = (
    kpi['Achievement_Percent']*0.35
    + kpi['Yield_Percent']*0.30
    + (100-kpi['Defect_Percent'])*0.15
    + (100-kpi['Downtime_Hours']*5)*0.20
)
kpi['Plant_Health_Score'] = kpi['Plant_Health_Score'].round(2)
kpi[['Date','Plant_Health_Score']].head()
```

```python
kpi[['Date', 'Plant_Health_Score']].head()
print(kpi.columns.tolist())
```

```python
def classify(score):

    if score>=95:
        return "Excellent"

    elif score>=90:
        return "Good"

    elif score>=85:
        return "Average"

    else:
        return "Needs Attention"

kpi['Plant_Status'] = kpi['Plant_Health_Score'].apply(classify)
kpi[['Date','Plant_Health_Score','Plant_Status']].head()
```

## Detect Operational Anomalies

```python
alerts = []
for _, row in kpi.iterrows():

    if row['Downtime_Hours'] > 3:
        alerts.append(
            f"{row['Date']} : High downtime ({row['Downtime_Hours']} hrs)"
        )

    if row['Yield_Percent'] < 92:
        alerts.append(
            f"{row['Date']} : Low Yield ({row['Yield_Percent']}%)"
        )

    if row['Defect_Percent'] > 2:
        alerts.append(
            f"{row['Date']} : High Defect ({row['Defect_Percent']}%)"
        )
alerts[:10]
```

```python
top_days = (
    kpi
    .sort_values(
        by='Plant_Health_Score',
        ascending=False
    )
    [['Date',
      'Production_Tonnes',
      'Plant_Health_Score',
      'Plant_Status']]
)

top_days.head(10)
```

```python
worst_days = (
    kpi
    .sort_values(by='Plant_Health_Score', ascending=True)
    [['Date',
      'Production_Tonnes',
      'Plant_Health_Score',
      'Plant_Status']]
)

worst_days.head(10)
```

```python
kpi['Month'] = pd.to_datetime(kpi['Date']).dt.to_period('M')

monthly_summary = (
    kpi
    .groupby('Month')
    .agg({
        'Production_Tonnes':'sum',
        'Revenue_INR':'sum',
        'Profit_per_Tonne_INR':'mean',
        'Yield_Percent':'mean',
        'Plant_Health_Score':'mean'
    })
    .round(2)
)

monthly_summary
```

# Exploratory Data Analysis & Business Insights

```python
# Daily Production Trend
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(14,6))

sns.lineplot(
    data=kpi,
    x='Date',
    y='Production_Tonnes',
    color='steelblue',
    linewidth=2
)

plt.title("Daily Steel Production Trend", fontsize=16)
plt.xlabel("Date")
plt.ylabel("Production (Tonnes)")
plt.xticks(rotation=45)
plt.grid(alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
# Production vs Target
plt.figure(figsize=(14,6))

plt.plot(kpi['Date'],
         kpi['Production_Tonnes'],
         label='Actual',
         linewidth=2)

plt.plot(kpi['Date'],
         kpi['Target_Tonnes'],
         label='Target',
         linestyle='--')

plt.legend()

plt.title("Actual vs Target Production")
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

```python
# Downtime Breakdown
downtime_totals = downtime.drop(columns=['Date','Total_Downtime_Hours']).sum()

plt.figure(figsize=(9,9))

plt.pie(
    downtime_totals,
    labels=downtime_totals.index,
    autopct='%1.1f%%',
    startangle=140
)

plt.title("Downtime Reason Distribution")

plt.show()
```

```python
# relationships between production, energy, costs, downtime, yield, and profit.
plt.figure(figsize=(12,8))

corr = kpi.select_dtypes(include='number').corr()

sns.heatmap(
    corr,
    cmap='RdBu_r',
    center=0,
    annot=False
)

plt.title("Correlation between Manufacturing KPIs")

plt.show()
```

```python
# Yield Distribution
plt.figure(figsize=(8,5))

sns.histplot(
    kpi['Yield_Percent'],
    bins=20,
    kde=True,
    color='green'
)

plt.title("Distribution of Plant Yield")

plt.show()
```

```python
plt.figure(figsize=(8,6))

sns.scatterplot(
    data=kpi,
    x='Downtime_Hours',
    y='Production_Tonnes',
    s=70
)

plt.title("Production vs Downtime")

plt.show()
```

```python
plt.figure(figsize=(14,6))

plt.plot(
    kpi['Date'],
    kpi['Plant_Health_Score'],
    color='darkgreen'
)

plt.title("Plant Health Score Over Time")

plt.ylabel("Health Score")

plt.xticks(rotation=45)

plt.tight_layout()

plt.show()
```

```python
# Profit vs Energy Consumption 
plt.figure(figsize=(8,6))

sns.regplot(
    data=kpi,
    x='Energy_Consumption_MWh',
    y='Profit_per_Tonne_INR',
    scatter_kws={'alpha':0.6},
    line_kws={'color':'red'}
)

plt.title("Profit per Tonne vs Energy Consumption")

plt.show()
```
